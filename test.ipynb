### Creat hugging_face_data csv files These csv files are having some ready made answers so lets keep them for tesing or evaluation purpose

```python
import pandas as pd
import os

# Assuming dataset is already loaded
# Example:
# from datasets import load_dataset
# dataset = load_dataset("your_dataset_name")

os.makedirs("data", exist_ok=True)

samples_rows = []
kpi_rows = []
anomaly_rows = []
statistics_rows = []
labels_rows = []
qna_rows = []

for sample_id, record in enumerate(dataset["full"]):

    # -----------------------------
    # 1. Main sample information
    # -----------------------------
    samples_rows.append({
        "sample_id": sample_id,
        "start_time": record.get("start_time"),
        "end_time": record.get("end_time"),
        "sampling_rate": record.get("sampling_rate"),
        "description": record.get("description")
    })

    # -----------------------------
    # 2. KPIs
    # -----------------------------
    kpis = record.get("KPIs")

    if isinstance(kpis, dict):
        for key, value in kpis.items():
            kpi_rows.append({
                "sample_id": sample_id,
                "kpi_name": key,
                "kpi_value": value
            })

    elif isinstance(kpis, list):
        for idx, item in enumerate(kpis):
            if isinstance(item, dict):
                row = {"sample_id": sample_id, "kpi_index": idx}
                row.update(item)
                kpi_rows.append(row)
            else:
                kpi_rows.append({
                    "sample_id": sample_id,
                    "kpi_index": idx,
                    "kpi_value": item
                })

    # -----------------------------
    # 3. Anomalies
    # -----------------------------
    anomalies = record.get("anomalies")

    if isinstance(anomalies, list):
        for idx, item in enumerate(anomalies):
            if isinstance(item, dict):
                row = {"sample_id": sample_id, "anomaly_index": idx}
                row.update(item)
                anomaly_rows.append(row)
            else:
                anomaly_rows.append({
                    "sample_id": sample_id,
                    "anomaly_index": idx,
                    "anomaly_value": item
                })

    elif isinstance(anomalies, dict):
        row = {"sample_id": sample_id}
        row.update(anomalies)
        anomaly_rows.append(row)

    # -----------------------------
    # 4. Statistics
    # -----------------------------
    statistics = record.get("statistics")

    if isinstance(statistics, dict):
        row = {"sample_id": sample_id}
        row.update(statistics)
        statistics_rows.append(row)

    elif isinstance(statistics, list):
        for idx, item in enumerate(statistics):
            if isinstance(item, dict):
                row = {"sample_id": sample_id, "statistics_index": idx}
                row.update(item)
                statistics_rows.append(row)

    # -----------------------------
    # 5. Labels
    # -----------------------------
    labels = record.get("labels")

    if isinstance(labels, dict):
        row = {"sample_id": sample_id}
        row.update(labels)
        labels_rows.append(row)

    elif isinstance(labels, list):
        for idx, item in enumerate(labels):
            labels_rows.append({
                "sample_id": sample_id,
                "label_index": idx,
                "label": item
            })

    else:
        labels_rows.append({
            "sample_id": sample_id,
            "label": labels
        })

    # -----------------------------
    # 6. QnA
    # -----------------------------
    qna = record.get("QnA")

    if isinstance(qna, list):
        for idx, item in enumerate(qna):
            if isinstance(item, dict):
                row = {"sample_id": sample_id, "qna_index": idx}
                row.update(item)
                qna_rows.append(row)
            else:
                qna_rows.append({
                    "sample_id": sample_id,
                    "qna_index": idx,
                    "qna_value": item
                })

    elif isinstance(qna, dict):
        row = {"sample_id": sample_id}
        row.update(qna)
        qna_rows.append(row)


# Convert to DataFrames
samples_df = pd.DataFrame(samples_rows)
kpis_df = pd.DataFrame(kpi_rows)
anomalies_df = pd.DataFrame(anomaly_rows)
statistics_df = pd.DataFrame(statistics_rows)
labels_df = pd.DataFrame(labels_rows)
qna_df = pd.DataFrame(qna_rows)


# Save CSV files
samples_df.to_csv("data/samples.csv", index=False)
kpis_df.to_csv("data/kpis.csv", index=False)
anomalies_df.to_csv("data/anomalies.csv", index=False)
statistics_df.to_csv("data/statistics.csv", index=False)
labels_df.to_csv("data/labels.csv", index=False)
qna_df.to_csv("data/qna.csv", index=False)

print("CSV files created successfully.")

print("\nSamples:")
print(samples_df.head())

print("\nKPIs:")
print(kpis_df.head())

print("\nAnomalies:")
print(anomalies_df.head())

print("\nStatistics:")
print(statistics_df.head())

print("\nLabels:")
print(labels_df.head())

print("\nQnA:")
print(qna_df.head())
```

### Generate actual telecom data from python script

#### cell_inventory.csv

```python
import pandas as pd
import random

random.seed(42)

cities = {
    "Bangalore": {
        "region": "South",
        "lat": 12.9716,
        "lon": 77.5946
    },
    "Hyderabad": {
        "region": "South",
        "lat": 17.3850,
        "lon": 78.4867
    },
    "Chennai": {
        "region": "South",
        "lat": 13.0827,
        "lon": 80.2707
    },
    "Mumbai": {
        "region": "West",
        "lat": 19.0760,
        "lon": 72.8777
    },
    "Delhi": {
        "region": "North",
        "lat": 28.6139,
        "lon": 77.2090
    }
}

rows = []

site_count = 40

for site_num in range(1, site_count + 1):

    site_id = f"SITE_{site_num:03d}"

    city = random.choice(list(cities.keys()))
    region = cities[city]["region"]

    base_lat = cities[city]["lat"]
    base_lon = cities[city]["lon"]

    # small random location variation around city
    latitude = base_lat + random.uniform(-0.08, 0.08)
    longitude = base_lon + random.uniform(-0.08, 0.08)

    # each site has 3 sectors
    for sector in range(1, 4):

        cell_id = f"CELL_{site_num:03d}_{sector}"

        technology = random.choices(
            ["4G", "5G"],
            weights=[0.65, 0.35]
        )[0]

        if technology == "4G":
            band = random.choice(["B3", "B8", "B40"])
            bandwidth = random.choice([10, 15, 20])

        else:
            band = "n78"
            bandwidth = random.choice([40, 60, 80, 100])

        azimuth = (sector - 1) * 120

        vendor = random.choices(
            ["Nokia", "Ericsson", "Samsung"],
            weights=[0.65, 0.25, 0.10]
        )[0]

        status = random.choices(
            ["ACTIVE", "LOCKED", "MAINTENANCE"],
            weights=[0.96, 0.02, 0.02]
        )[0]

        rows.append({
            "cell_id": cell_id,
            "site_id": site_id,
            "region": region,
            "city": city,
            "technology": technology,
            "band": band,
            "bandwidth_mhz": bandwidth,
            "vendor": vendor,
            "latitude": round(latitude, 6),
            "longitude": round(longitude, 6),
            "azimuth": azimuth,
            "cell_status": status
        })


df = pd.DataFrame(rows)

df.to_csv(
    "data/telecom_data/cell_inventory.csv",
    index=False
)

print("cell_inventory.csv created successfully")

print("\nTotal cells:", len(df))

print("\nTechnology distribution:")
print(df["technology"].value_counts())

print("\nCity distribution:")
print(df["city"].value_counts())

print("\nSample rows:")
print(df.head(10))
```

#### create real type telecom data

```python
import pandas as pd
import numpy as np
import random
from pathlib import Path
from datetime import timedelta

# ---------------------------------------------------------
# Configuration
# ---------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

DATA_DIR = Path("data/telecom_data")
EVAL_DIR = DATA_DIR / "evaluation"

DATA_DIR.mkdir(parents=True, exist_ok=True)
EVAL_DIR.mkdir(parents=True, exist_ok=True)

START_DATE = pd.Timestamp("2026-08-01 00:00:00")
DAYS = 30
FREQUENCY = "15min"


# ---------------------------------------------------------
# Load inventory
# ---------------------------------------------------------

inventory = pd.read_csv(DATA_DIR / "cell_inventory.csv")

print("Inventory loaded")
print("Cells:", len(inventory))


# ---------------------------------------------------------
# Generate timestamps
# ---------------------------------------------------------

timestamps = pd.date_range(
    start=START_DATE,
    periods=DAYS * 24 * 4,
    freq=FREQUENCY
)

print("Time points per cell:", len(timestamps))


# ---------------------------------------------------------
# Traffic profile
# ---------------------------------------------------------

def traffic_factor(hour):
    """
    Simulates realistic daily traffic pattern.
    """

    if 0 <= hour < 6:
        return 0.25

    elif 6 <= hour < 9:
        return 0.55

    elif 9 <= hour < 12:
        return 0.75

    elif 12 <= hour < 17:
        return 0.85

    elif 17 <= hour < 22:
        return 1.0

    else:
        return 0.50


# ---------------------------------------------------------
# Create degradation scenarios
# ---------------------------------------------------------

scenario_types = [
    "CONGESTION",
    "RADIO_INTERFERENCE",
    "TRANSPORT_DEGRADATION",
    "CELL_OUTAGE",
    "CONFIGURATION_REGRESSION"
]

scenario_counts = {
    "CONGESTION": 18,
    "RADIO_INTERFERENCE": 15,
    "TRANSPORT_DEGRADATION": 12,
    "CELL_OUTAGE": 8,
    "CONFIGURATION_REGRESSION": 10
}

scenarios = []

scenario_id = 1

all_cells = inventory["cell_id"].tolist()

for scenario_type, count in scenario_counts.items():

    for _ in range(count):

        cell_id = random.choice(all_cells)

        # avoid first and last day
        random_day = random.randint(1, DAYS - 2)

        start_hour = random.randint(7, 21)

        start_time = (
            START_DATE
            + timedelta(days=random_day)
            + timedelta(hours=start_hour)
        )

        duration_hours = random.choice([1, 2, 3, 4])

        end_time = start_time + timedelta(hours=duration_hours)

        scenarios.append({
            "scenario_id": f"SCN_{scenario_id:04d}",
            "cell_id": cell_id,
            "scenario_type": scenario_type,
            "start_time": start_time,
            "end_time": end_time
        })

        scenario_id += 1


scenario_df = pd.DataFrame(scenarios)

scenario_df.to_csv(
    EVAL_DIR / "scenario_truth.csv",
    index=False
)

print("\nScenarios created:", len(scenario_df))


# ---------------------------------------------------------
# Helper: find active scenario
# ---------------------------------------------------------

scenario_lookup = {}

for _, scenario in scenario_df.iterrows():

    cell = scenario["cell_id"]

    if cell not in scenario_lookup:
        scenario_lookup[cell] = []

    scenario_lookup[cell].append(scenario)


# ---------------------------------------------------------
# KPI GENERATION
# ---------------------------------------------------------

kpi_rows = []

for _, cell in inventory.iterrows():

    cell_id = cell["cell_id"]
    site_id = cell["site_id"]
    technology = cell["technology"]

    # Different cells have slightly different normal behavior
    base_users = random.randint(80, 250)

    base_rsrp = random.uniform(-95, -78)

    for ts in timestamps:

        tf = traffic_factor(ts.hour)

        # -----------------------------------------------
        # Normal KPI values
        # -----------------------------------------------

        active_users = max(
            5,
            int(
                base_users
                * tf
                * np.random.normal(1, 0.08)
            )
        )

        dl_prb = np.clip(
            25 + (tf * 55) + np.random.normal(0, 5),
            5,
            95
        )

        ul_prb = np.clip(
            15 + (tf * 40) + np.random.normal(0, 5),
            3,
            90
        )

        rrc_success = np.clip(
            np.random.normal(98.8, 0.5),
            94,
            100
        )

        call_drop_rate = np.clip(
            np.random.normal(0.8, 0.25),
            0.1,
            3
        )

        handover_success = np.clip(
            np.random.normal(97.5, 1),
            90,
            100
        )

        availability = np.clip(
            np.random.normal(99.98, 0.02),
            99.5,
            100
        )

        latency = np.clip(
            np.random.normal(18, 4),
            5,
            50
        )

        packet_loss = np.clip(
            np.random.normal(0.3, 0.12),
            0,
            2
        )

        rsrp = np.clip(
            np.random.normal(base_rsrp, 3),
            -120,
            -65
        )

        sinr = np.clip(
            np.random.normal(18, 4),
            -5,
            30
        )

        # Throughput decreases as PRB increases
        dl_throughput = max(
            5,
            np.random.normal(
                250 if technology == "5G" else 70,
                15
            )
            * (1 - dl_prb / 160)
        )

        ul_throughput = max(
            2,
            np.random.normal(
                45 if technology == "5G" else 20,
                5
            )
            * (1 - ul_prb / 180)
        )

        erab_success = np.nan
        pdu_session_success = np.nan

        if technology == "4G":

            erab_success = np.clip(
                np.random.normal(98.2, 0.7),
                92,
                100
            )

        else:

            pdu_session_success = np.clip(
                np.random.normal(98.5, 0.6),
                92,
                100
            )

        # -----------------------------------------------
        # Inject abnormal behavior
        # -----------------------------------------------

        active_scenarios = scenario_lookup.get(cell_id, [])

        for scenario in active_scenarios:

            if (
                scenario["start_time"]
                <= ts
                < scenario["end_time"]
            ):

                scenario_type = scenario["scenario_type"]

                # ---------------------------------------
                # Congestion
                # ---------------------------------------

                if scenario_type == "CONGESTION":

                    dl_prb = np.clip(
                        np.random.normal(96, 2),
                        88,
                        100
                    )

                    ul_prb = np.clip(
                        np.random.normal(88, 4),
                        70,
                        100
                    )

                    active_users = int(
                        active_users
                        * random.uniform(1.6, 2.2)
                    )

                    dl_throughput *= random.uniform(
                        0.35,
                        0.60
                    )

                    latency *= random.uniform(
                        1.8,
                        2.8
                    )

                    call_drop_rate *= random.uniform(
                        2,
                        4
                    )

                # ---------------------------------------
                # Radio interference
                # ---------------------------------------

                elif scenario_type == "RADIO_INTERFERENCE":

                    rsrp -= random.uniform(8, 18)

                    sinr -= random.uniform(10, 18)

                    call_drop_rate *= random.uniform(
                        3,
                        6
                    )

                    handover_success -= random.uniform(
                        5,
                        12
                    )

                    rrc_success -= random.uniform(
                        2,
                        6
                    )

                    dl_throughput *= random.uniform(
                        0.4,
                        0.7
                    )

                # ---------------------------------------
                # Transport degradation
                # ---------------------------------------

                elif scenario_type == "TRANSPORT_DEGRADATION":

                    latency *= random.uniform(
                        3,
                        6
                    )

                    packet_loss *= random.uniform(
                        8,
                        20
                    )

                    dl_throughput *= random.uniform(
                        0.35,
                        0.65
                    )

                    ul_throughput *= random.uniform(
                        0.4,
                        0.7
                    )

                # ---------------------------------------
                # Cell outage
                # ---------------------------------------

                elif scenario_type == "CELL_OUTAGE":

                    availability = random.uniform(
                        0,
                        20
                    )

                    active_users = random.randint(
                        0,
                        5
                    )

                    dl_throughput = 0

                    ul_throughput = 0

                    rrc_success = random.uniform(
                        0,
                        30
                    )

                    if technology == "4G":
                        erab_success = random.uniform(
                            0,
                            30
                        )

                    else:
                        pdu_session_success = random.uniform(
                            0,
                            30
                        )

                # ---------------------------------------
                # Configuration regression
                # ---------------------------------------

                elif scenario_type == "CONFIGURATION_REGRESSION":

                    handover_success -= random.uniform(
                        8,
                        18
                    )

                    call_drop_rate *= random.uniform(
                        2,
                        4
                    )

                    rrc_success -= random.uniform(
                        1,
                        4
                    )

        # -----------------------------------------------
        # Clamp values
        # -----------------------------------------------

        rrc_success = np.clip(
            rrc_success,
            0,
            100
        )

        handover_success = np.clip(
            handover_success,
            0,
            100
        )

        call_drop_rate = np.clip(
            call_drop_rate,
            0,
            100
        )

        latency = np.clip(
            latency,
            0,
            500
        )

        packet_loss = np.clip(
            packet_loss,
            0,
            100
        )

        sinr = np.clip(
            sinr,
            -20,
            40
        )

        rsrp = np.clip(
            rsrp,
            -140,
            -50
        )

        # -----------------------------------------------
        # Store KPI row
        # -----------------------------------------------

        kpi_rows.append({

            "timestamp": ts,

            "cell_id": cell_id,

            "site_id": site_id,

            "technology": technology,

            "dl_prb_utilization": round(dl_prb, 2),

            "ul_prb_utilization": round(ul_prb, 2),

            "rrc_setup_success_rate": round(
                rrc_success,
                2
            ),

            "erab_setup_success_rate": (
                round(erab_success, 2)
                if not pd.isna(erab_success)
                else np.nan
            ),

            "pdu_session_success_rate": (
                round(pdu_session_success, 2)
                if not pd.isna(pdu_session_success)
                else np.nan
            ),

            "call_drop_rate": round(
                call_drop_rate,
                2
            ),

            "handover_success_rate": round(
                handover_success,
                2
            ),

            "dl_throughput_mbps": round(
                dl_throughput,
                2
            ),

            "ul_throughput_mbps": round(
                ul_throughput,
                2
            ),

            "active_users": active_users,

            "availability_pct": round(
                availability,
                3
            ),

            "latency_ms": round(
                latency,
                2
            ),

            "packet_loss_pct": round(
                packet_loss,
                2
            ),

            "rsrp_dbm": round(
                rsrp,
                2
            ),

            "sinr_db": round(
                sinr,
                2
            )
        })


kpi_df = pd.DataFrame(kpi_rows)

kpi_df.to_csv(
    DATA_DIR / "cell_kpi.csv",
    index=False
)

print("\nKPI rows:", len(kpi_df))


# ---------------------------------------------------------
# ALARM GENERATION
# ---------------------------------------------------------

alarm_rows = []

alarm_id = 1

background_alarms = [
    ("MINOR_VSWR_WARNING", "MINOR"),
    ("SYNCHRONIZATION_WARNING", "MINOR"),
    ("TRANSPORT_LINK_FLAP", "WARNING"),
    ("HIGH_TEMPERATURE", "MINOR"),
    ("POWER_SUPPLY_WARNING", "WARNING")
]

# background noise alarms
for _, cell in inventory.iterrows():

    number_of_alarms = np.random.poisson(10)

    for _ in range(number_of_alarms):

        alarm_name, severity = random.choice(
            background_alarms
        )

        start_offset = random.randint(
            0,
            DAYS * 24 * 60 - 1
        )

        alarm_time = (
            START_DATE
            + timedelta(minutes=start_offset)
        )

        duration_minutes = random.choice(
            [5, 10, 15, 30, 60, 120]
        )

        clear_time = (
            alarm_time
            + timedelta(minutes=duration_minutes)
        )

        alarm_rows.append({

            "alarm_id": f"ALM_{alarm_id:06d}",

            "timestamp": alarm_time,

            "clear_time": clear_time,

            "cell_id": cell["cell_id"],

            "site_id": cell["site_id"],

            "alarm_name": alarm_name,

            "severity": severity,

            "status": "CLEARED"
        })

        alarm_id += 1


# scenario-linked alarms
scenario_alarm_mapping = {

    "CONGESTION":
        ("HIGH_RESOURCE_UTILIZATION", "MAJOR"),

    "RADIO_INTERFERENCE":
        ("RADIO_QUALITY_DEGRADATION", "MAJOR"),

    "TRANSPORT_DEGRADATION":
        ("TRANSPORT_PACKET_LOSS", "MAJOR"),

    "CELL_OUTAGE":
        ("CELL_UNAVAILABLE", "CRITICAL"),

    "CONFIGURATION_REGRESSION":
        ("HANDOVER_FAILURE_RATE_HIGH", "MAJOR")
}


for _, scenario in scenario_df.iterrows():

    cell_info = inventory[
        inventory["cell_id"]
        == scenario["cell_id"]
    ].iloc[0]

    alarm_name, severity = (
        scenario_alarm_mapping[
            scenario["scenario_type"]
        ]
    )

    alarm_rows.append({

        "alarm_id": f"ALM_{alarm_id:06d}",

        "timestamp":
            scenario["start_time"]
            + timedelta(minutes=random.choice([0, 5, 10, 15])),

        "clear_time":
            scenario["end_time"],

        "cell_id":
            scenario["cell_id"],

        "site_id":
            cell_info["site_id"],

        "alarm_name":
            alarm_name,

        "severity":
            severity,

        "status":
            "CLEARED"
    })

    alarm_id += 1


alarms_df = pd.DataFrame(alarm_rows)

alarms_df = alarms_df.sort_values(
    "timestamp"
)

alarms_df.to_csv(
    DATA_DIR / "alarms.csv",
    index=False
)

print("Alarm rows:", len(alarms_df))


# ---------------------------------------------------------
# CONFIGURATION CHANGES
# ---------------------------------------------------------

config_rows = []

change_id = 1

parameters = [
    "handover_margin_db",
    "cell_reselection_offset_db",
    "tx_power_dbm",
    "max_connected_users",
    "antenna_tilt_deg",
    "qrxlevmin_dbm"
]


for _, cell in inventory.iterrows():

    number_changes = random.randint(1, 4)

    for _ in range(number_changes):

        parameter = random.choice(parameters)

        change_time = (
            START_DATE
            + timedelta(
                minutes=random.randint(
                    0,
                    DAYS * 24 * 60 - 1
                )
            )
        )

        old_value = round(
            random.uniform(1, 20),
            2
        )

        new_value = round(
            old_value
            + random.uniform(-3, 3),
            2
        )

        config_rows.append({

            "change_id":
                f"CHG_{change_id:05d}",

            "timestamp":
                change_time,

            "cell_id":
                cell["cell_id"],

            "site_id":
                cell["site_id"],

            "parameter_name":
                parameter,

            "old_value":
                old_value,

            "new_value":
                new_value,

            "changed_by":
                random.choice([
                    "OSS_AUTOMATION",
                    "RAN_ENGINEER",
                    "SON_PLATFORM"
                ])
        })

        change_id += 1


# Add configuration changes before config regression scenarios
for _, scenario in scenario_df[
    scenario_df["scenario_type"]
    == "CONFIGURATION_REGRESSION"
].iterrows():

    cell_info = inventory[
        inventory["cell_id"]
        == scenario["cell_id"]
    ].iloc[0]

    config_rows.append({

        "change_id":
            f"CHG_{change_id:05d}",

        "timestamp":
            scenario["start_time"]
            - timedelta(minutes=30),

        "cell_id":
            scenario["cell_id"],

        "site_id":
            cell_info["site_id"],

        "parameter_name":
            "handover_margin_db",

        "old_value":
            3.0,

        "new_value":
            8.0,

        "changed_by":
            "RAN_ENGINEER"
    })

    change_id += 1


config_df = pd.DataFrame(config_rows)

config_df = config_df.sort_values(
    "timestamp"
)

config_df.to_csv(
    DATA_DIR / "configuration_changes.csv",
    index=False
)

print("Configuration changes:", len(config_df))


# ---------------------------------------------------------
# TICKET GENERATION
# ---------------------------------------------------------

ticket_rows = []

ticket_id = 1

issue_messages = {

    "CONGESTION":
        "Users reporting slow data throughput and degraded service.",

    "RADIO_INTERFERENCE":
        "Users experiencing intermittent call drops and poor radio quality.",

    "TRANSPORT_DEGRADATION":
        "Customer traffic experiencing high latency and intermittent packet loss.",

    "CELL_OUTAGE":
        "Multiple users reporting complete service loss in the affected area.",

    "CONFIGURATION_REGRESSION":
        "Call mobility performance degraded after recent network activity."
}


for _, scenario in scenario_df.iterrows():

    # not every scenario results in customer ticket
    if random.random() < 0.75:

        created_time = (
            scenario["start_time"]
            + timedelta(
                minutes=random.randint(
                    10,
                    90
                )
            )
        )

        ticket_rows.append({

            "ticket_id":
                f"TKT_{ticket_id:05d}",

            "created_time":
                created_time,

            "cell_id":
                scenario["cell_id"],

            "priority":
                random.choice([
                    "P1",
                    "P2",
                    "P3"
                ]),

            "source":
                random.choice([
                    "CUSTOMER_COMPLAINT",
                    "NOC",
                    "SERVICE_DESK"
                ]),

            "issue_summary":
                issue_messages[
                    scenario["scenario_type"]
                ],

            "status":
                random.choice([
                    "CLOSED",
                    "RESOLVED"
                ])
        })

        ticket_id += 1


# Add unrelated tickets/noise
generic_ticket_messages = [
    "User reports intermittent mobile data performance.",
    "Customer experiencing low throughput during peak hours.",
    "Voice quality degradation reported by multiple users.",
    "Intermittent service degradation observed.",
    "Customer reports connectivity issue."
]

for _ in range(200):

    cell = inventory.sample(1).iloc[0]

    created_time = (
        START_DATE
        + timedelta(
            minutes=random.randint(
                0,
                DAYS * 24 * 60 - 1
            )
        )
    )

    ticket_rows.append({

        "ticket_id":
            f"TKT_{ticket_id:05d}",

        "created_time":
            created_time,

        "cell_id":
            cell["cell_id"],

        "priority":
            random.choice([
                "P2",
                "P3",
                "P4"
            ]),

        "source":
            random.choice([
                "CUSTOMER_COMPLAINT",
                "SERVICE_DESK"
            ]),

        "issue_summary":
            random.choice(
                generic_ticket_messages
            ),

        "status":
            random.choice([
                "CLOSED",
                "RESOLVED"
            ])
    })

    ticket_id += 1


tickets_df = pd.DataFrame(ticket_rows)

tickets_df = tickets_df.sort_values(
    "created_time"
)

tickets_df.to_csv(
    DATA_DIR / "tickets.csv",
    index=False
)

print("Tickets:", len(tickets_df))


# ---------------------------------------------------------
# Final summary
# ---------------------------------------------------------

print("\n----------------------------------------")
print("TELECOM DATA GENERATION COMPLETE")
print("----------------------------------------")

print(
    "Inventory:",
    len(inventory)
)

print(
    "KPI rows:",
    len(kpi_df)
)

print(
    "Alarm rows:",
    len(alarms_df)
)

print(
    "Config changes:",
    len(config_df)
)

print(
    "Tickets:",
    len(tickets_df)
)

print(
    "Hidden RCA scenarios:",
    len(scenario_df)
)

print("----------------------------------------")
```
